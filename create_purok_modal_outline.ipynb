# Create Purok Modal Map Outline

A Jupyter notebook outline for implementing a Create Purok modal in a Blade file with a map limited to the assigned barangay range, custom range support, and refresh behavior on modal close.

## 1. Define the Create Purok Modal Trigger in `bhw_workers.blade.php`
Create the button or link that opens the Create Purok modal and wire it to the Blade view structure.

## 2. Load Barangay Boundary Data and Map Dependencies
Prepare the barangay boundary coordinates or GeoJSON data and include the required map library assets.

## 3. Initialize the Map Inside the Modal
Render the map when the modal opens and center it on the assigned barangay area.

## 4. Restrict the Map View to the Assigned Barangay Range
Limit panning, zooming, and selectable areas so the user cannot select outside the assigned barangay boundary.

## 5. Add Support for a Custom Range Rule
Allow custom range configuration based on local LGU requirements and apply it to the map overlay or boundary logic.

## 6. Handle Modal Close, Reset, and Refresh Behavior
Reset the map state when the modal closes and refresh the map each time the modal is reopened.

## 7. Validate and Submit the Selected Purok Range
Validate the final selected range before saving and send the data to the backend for storage.
